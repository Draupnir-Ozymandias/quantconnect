# QCRL ResearchTools v2

Use this notebook after at least one backtest has saved normalized experiment records through `ExperimentStore`.

```python
from experiment_store import ExperimentStore
from research_tools import ResearchTools

tools = ResearchTools(qb.object_store)
records = tools.load(refresh=True)

print(f'Loaded {len(records)} experiment records')
```

## Top runs

```python
top_runs = tools.top(sort_by='risk_adjusted_score', n=10)
print(tools.show(top_runs))
```

## Best surviving runs

```python
best_survivors = tools.best_survivors(sort_by='risk_adjusted_score', n=10)
print(tools.show(best_survivors))
```

## Select a cohort

```python
btc_1h = tools.select(
    coin='BTCUSD',
    timeframe='1h',
    stake_mode='martingale'
)

print(f'BTCUSD 1h martingale cohort: {len(btc_1h)} records')
print(tools.show(tools.top(btc_1h, n=10)))
```

## Filter with numeric constraints

```python
safe_profitable = tools.select(
    ruined=False,
    net_profit__gt=0,
    max_drawdown__lte=5000
)

print(f'Safe/profitable cohort: {len(safe_profitable)} records')
print(tools.show(tools.top(safe_profitable, n=10)))
```

## Cohort summaries

```python
summary_by_timeframe = tools.cohort_summary(group_by='timeframe')
print(tools.format_table(summary_by_timeframe))
```

```python
summary_by_entry_filter = tools.cohort_summary(
    group_by=['entry_model', 'filter_model']
)
print(tools.format_table(summary_by_entry_filter))
```

## Compare specific runs

```python
# Replace these with real run IDs from the top-runs table.
run_ids = [r['run_id'] for r in top_runs[:2]]

print(tools.format_table(tools.compare(run_ids)))
```

## Full report for one run

```python
if len(top_runs) > 0:
    run_id = top_runs[0]['run_id']
    report = tools.load_report(run_id)
    print(run_id)
    print(report.keys() if report is not None else 'Report not found')
```
